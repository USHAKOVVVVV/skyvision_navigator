```python
from ultralytics import YOLO

# Загрузка предобученной модели
model = YOLO('Моделюси/yolov8n_gpu_simple_1.pt')  # для сегментации
```

```python
results = model.train(
    data='data.yaml',
    epochs=30,          # Эпохи
    
    imgsz=640, # Размер изображеньки ставим 640, т.к. в этом формате размечаем
    
    batch=2,            # По скок изображенек берем за прогон
    
    patience=15,    #Терпение, хз че это значит ?

    device='cuda',      # Используем CPU, если нет видеокарты, если есть, то ставим 'cuda'

    workers=2,          # Минимум workers
    
    lr0=0.005,   # Лернинг рэйт ?

    optimizer='AdamW',   # Adam - легкий оптимизатор, AdamW - Жестче

    # weight_decay = 0.0005,   # Регуляризация ?
    
    name='yolov8n_gpu_updgrade_1',
    
    exist_ok=True
)
```

```python
from ultralytics import YOLO
import os

# Проверка структуры датасета
def check_dataset_structure(dataset_path):
    print("Проверка структуры датасета...")
    
    required_dirs = [
        'train/images',
        'valid/images', 
        'train/labels',
        'valid/labels'
    ]
    
    for dir_path in required_dirs:
        full_path = os.path.join(dataset_path, dir_path)
        if not os.path.exists(full_path):
            print(f"❌ Отсутствует: {full_path}")
        else:
            print(f"✅ Найдено: {full_path}")
            
    # Проверка файлов labels
    labels_train_path = os.path.join(dataset_path, 'train/labels')
    if os.path.exists(labels_train_path):
        label_files = os.listdir(labels_train_path)
        if label_files:
            sample_label = os.path.join(labels_train_path, label_files[0])
            with open(sample_label, 'r') as f:
                content = f.read().strip()
                print(f"Пример файла разметки: {content}")
                
                # Проверка формата
                if content:
                    parts = content.split()
                    if len(parts) > 5:
                        print("✅ Формат: Сегментация (полигоны)")
                    else:
                        print("❌ Формат: Детекция (bbox), а нужна сегментация!")

# Проверка
dataset_path = ""  # укажите ваш путь
check_dataset_structure(dataset_path)
```

```python
model_path = 'runs/segment/yolov8n_gpu_upgrade_1/weights/best.pt'
model = YOLO(model_path)
```

```python
# Валидация на тестовых данных
results = model.val(
    data='data.yaml',
    split='test',    # используем ТЕСТОВЫЕ данные
    conf=0.25, # уровень достоверности
    iou=0.3, #максимальный порог пересечения объектов
    device='cpu'
)

print(f"📊 Результаты оценки:")
print(f"mAP50: {results.box.map50:.3f}")

# print(results.box)

# Если mAP50 < 0.6 → нужно улучшать модель
# Если mAP50 > 0.8 → отличный результат!
```
